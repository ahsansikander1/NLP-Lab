# Task 1

## Task 1A

```python
pdf_file = r"C:\Users\Knl-Hunter\Downloads\Computer_Networking_Chapter_1_Complete_Study_Guide.pdf"

reader = PdfReader(pdf_file)
pages_text = []

for page in reader.pages:
    text = page.extract_text() or ""
    pages_text.append(text)

print("Total pages extracted:", len(pages_text))
```

## Task 1B

```python
all_text = " ".join(pages_text)
clean_text = re.sub(r"[^A-Za-z\s]", " ", all_text)
tokens = clean_text.lower().split()

print("Total token count:", len(tokens))
```

## Task 1C

```python
common_stopwords = {
    "the", "is", "a", "an", "and", "or", "to", "of", "in", "on",
    "for", "with", "as", "by", "at", "from", "this", "that", "it", "be",
    "are", "was", "were", "has", "have"
}

filtered_tokens = [word for word in tokens if word not in common_stopwords]
word_frequency = Counter(filtered_tokens)

for word, count in word_frequency.most_common(10):
    print(word, count)
```

## Task 1D

```python
keyword = input("Enter keyword: ").lower()
found = False

for page_number, page_text in enumerate(pages_text, start=1):
    sentences = re.split(r"(?<=[.!?])\s+", page_text)
    for sentence in sentences:
        if keyword in sentence.lower():
            print("Page", page_number, ":", sentence.strip())
            found = True

if not found:
    print("No matching sentence found.")
```

# Task 2

## Task 2A

```python
pdf_file_2 =  r"C:\Users\Knl-Hunter\Downloads\Computer_Networking_Chapter_1_Complete_Study_Guide.pdf"

reader_2 = PdfReader(pdf_file_2)
pages_data = []

for page_number, page in enumerate(reader_2.pages, start=1):
    raw_text = page.extract_text() or ""
    words = raw_text.split()
    char_count = len(raw_text.replace(" ", ""))

    pages_data.append({
        "page_number": page_number,
        "raw_text": raw_text,
        "word_count": len(words),
        "char_count": char_count
    })

print(pages_data[:2])
```

## Task 2B

```python
highest_page = max(pages_data, key=lambda x: x["word_count"])
non_zero_pages = [page for page in pages_data if page["word_count"] > 0]

print("Highest word count page:", highest_page["page_number"])
print(highest_page["raw_text"])

if non_zero_pages:
    lowest_page = min(non_zero_pages, key=lambda x: x["word_count"])
    print("Lowest non-zero word count page:", lowest_page["page_number"])
    print("Word count:", lowest_page["word_count"])
else:
    print("No page has a non-zero word count.")

average_word_count = sum(page["word_count"] for page in pages_data) / len(pages_data)
print("Average word count:", round(average_word_count, 2))
```

## Task 2C

```python
class PhraseNotFoundError(Exception):
    pass

search_phrase = input("Enter search phrase: ").lower()

try:
    matches = []

    for page in pages_data:
        count = page["raw_text"].lower().count(search_phrase)
        if count > 0:
            matches.append((page["page_number"], count))

    if not matches:
        raise PhraseNotFoundError("Phrase not found in any page.")

    for page_number, count in matches:
        print("Page", page_number, ":", count, "time(s)")

except PhraseNotFoundError as error:
    print(error)
```

## Task 2D

```python
with open("pdf_summary.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["page_number", "word_count", "char_count"])

    for page in pages_data:
        writer.writerow([
            page["page_number"],
            page["word_count"],
            page["char_count"]
        ])

with open("pdf_summary.csv", "r", encoding="utf-8") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

## Task 3A

```python
pdf_file_3 =  r"C:\Users\Knl-Hunter\Downloads\Computer_Networking_Chapter_1_Complete_Study_Guide.pdf"

reader_3 = PdfReader(pdf_file_3)
text = ""

for page in reader_3.pages:
    text += (page.extract_text() or "") + " "

before_count = len(text)

cleaned_text = re.sub(r"https?://\S+|www\.\S+", " ", text)
cleaned_text = re.sub(r"\d+", " ", cleaned_text)
cleaned_text = re.sub(r"[^A-Za-z. ]", " ", cleaned_text)
cleaned_text = re.sub(r" +", " ", cleaned_text)

after_count = len(cleaned_text)

print("Character count before cleaning:", before_count)
print("Character count after cleaning:", after_count)
```

## Task 3B

```python
cleaned_text = cleaned_text.lower()

sentences_before = re.split(r"[.!?]", cleaned_text)
sentences = [sentence.strip() for sentence in sentences_before]
sentences = [sentence for sentence in sentences if sentence]

print("Sentence count before removing empty strings:", len(sentences_before))
print("Sentence count after removing empty strings:", len(sentences))
```

## Task 3C

```python
try:
    english_stopwords = set(stopwords.words("english"))
except LookupError:
    nltk.download("stopwords")
    english_stopwords = set(stopwords.words("english"))

filtered_sentences = []

for sentence in sentences:
    words = sentence.split()
    remaining_words = [word for word in words if word not in english_stopwords]

    if len(remaining_words) >= 5:
        filtered_sentences.append(" ".join(remaining_words))

print("Sentence count before filter:", len(sentences))
print("Sentence count after filter:", len(filtered_sentences))

for sentence in filtered_sentences[:3]:
    print(sentence)
```

# Task 4

## Task 4A

```python
docx_file =  r"C:\Users\Knl-Hunter\Downloads\NLP Assignment 01.docx"

doc = Document(docx_file)
paragraphs_data = []

for paragraph in doc.paragraphs:
    paragraphs_data.append({
        "text": paragraph.text,
        "style": paragraph.style.name
    })

unique_styles = sorted(set(item["style"] for item in paragraphs_data))

print("Total paragraph count:", len(paragraphs_data))
print("Unique styles:", unique_styles)
```

## Task 4B

```python
groups = {
    "Headings": [],
    "Body": [],
    "Other/Empty": []
}

for item in paragraphs_data:
    text = item["text"]
    style = item["style"]

    if text.strip() == "":
        groups["Other/Empty"].append(item)
    elif style.startswith("Heading"):
        groups["Headings"].append(item)
    elif style == "Normal" or "Body" in style:
        groups["Body"].append(item)
    else:
        groups["Other/Empty"].append(item)

for group_name, items in groups.items():
    print(group_name, ":", len(items))
```

## Task 4C

```python
body_stats = []

for item in groups["Body"]:
    words = item["text"].split()

    if len(words) > 5:
        body_stats.append({
            "text": item["text"],
            "word_count": len(words),
            "char_count": len(item["text"].replace(" ", ""))
        })

if body_stats:
    highest_body = max(body_stats, key=lambda x: x["word_count"])
    lowest_body = min(body_stats, key=lambda x: x["word_count"])
    mean_words = sum(item["word_count"] for item in body_stats) / len(body_stats)

    print("Highest word count paragraph:")
    print(highest_body)

    print("Lowest word count paragraph above 5 words:")
    print(lowest_body)

    print("Mean word count:", round(mean_words, 2))
else:
    print("No qualifying body paragraphs found.")
```

## Task 4D

```python
style_dictionary = {}

for item in paragraphs_data:
    style = item["style"]

    if style not in style_dictionary:
        style_dictionary[style] = {
            "count": 0,
            "texts": []
        }

    style_dictionary[style]["count"] += 1
    style_dictionary[style]["texts"].append(item["text"])

sorted_styles = dict(
    sorted(
        style_dictionary.items(),
        key=lambda x: x[1]["count"],
        reverse=True
    )
)

for style in sorted_styles:
    if sorted_styles[style]["count"] > 2:
        sorted_styles[style]["texts"] = sorted_styles[style]["texts"][:2]

pprint.pprint(sorted_styles)
```

# Task 5

## Task 5A

```python
table_docx_file =  r"C:\Users\Knl-Hunter\Downloads\tables.docx"

table_doc = Document(table_docx_file)
tables_data = []

for table in table_doc.tables:
    table_data = []

    for row in table.rows:
        row_data = [cell.text for cell in row.cells]
        table_data.append(row_data)

    tables_data.append(table_data)

print("Total tables found:", len(tables_data))

for number, table_data in enumerate(tables_data, start=1):
    rows = len(table_data)
    columns = len(table_data[0]) if table_data else 0
    print("Table", number, ":", rows, "x", columns)
```

## Task 5B

```python
if tables_data:
    first_table = tables_data[0]
    headers = first_table[0]
    data_rows = first_table[1:]

    df = pd.DataFrame(data_rows, columns=headers)
    print(df.to_string(index=False))

    empty_counts = {}

    for column in df.columns:
        empty_counts[column] = sum(
            value is None or str(value).strip() == ""
            for value in df[column]
        )

    print("Empty or None cells per column:")
    print(empty_counts)
else:
    print("No table found.")
```

## Task 5C

```python
pre_table_paragraphs = []

for child in table_doc.element.body:
    if child.tag.endswith("}tbl"):
        break

    if child.tag.endswith("}p"):
        paragraph = Paragraph(child, table_doc)
        pre_table_paragraphs.append(paragraph.text)

print("Pre-table paragraph count:", len(pre_table_paragraphs))
```

## Task 5D

```python
with open("docx_extracted.txt", "w", encoding="utf-8") as file:
    file.write("--- PRE-TABLE PARAGRAPHS ---\n")

    for paragraph in pre_table_paragraphs:
        file.write(paragraph + "\n")

    file.write("--- TABLE DATA ---\n")

    if tables_data:
        for row in tables_data[0]:
            file.write(" | ".join(row) + "\n")

print("docx_extracted.txt created.")
```

# Task 6

## Task 6A

```python
document_a_file =  r"C:\Users\Knl-Hunter\Downloads\NLP Assignment 01.docx"
document_b_file =  r"C:\Users\Knl-Hunter\Downloads\Data_Visualization_Critique_and_Redesign.docx"

document_a = Document(document_a_file)
document_b = Document(document_b_file)

headings_a = [
    paragraph.text.strip()
    for paragraph in document_a.paragraphs
    if paragraph.style.name.startswith("Heading") and paragraph.text.strip()
]

headings_b = [
    paragraph.text.strip()
    for paragraph in document_b.paragraphs
    if paragraph.style.name.startswith("Heading") and paragraph.text.strip()
]

set_a = set(headings_a)
set_b = set(headings_b)

print("Common headings:", set_a & set_b)
print("Only in Document A:", set_a - set_b)
print("Only in Document B:", set_b - set_a)
```

## Task 6B

```python
normal_a = [
    paragraph.text.strip()
    for paragraph in document_a.paragraphs
    if paragraph.style.name == "Normal" and paragraph.text.strip()
]

normal_b = [
    paragraph.text.strip()
    for paragraph in document_b.paragraphs
    if paragraph.style.name == "Normal" and paragraph.text.strip()
]

merged_paragraphs = normal_a + normal_b

unique_paragraphs = []
seen = set()

for paragraph in merged_paragraphs:
    key = paragraph.strip().lower()

    if key not in seen:
        seen.add(key)
        unique_paragraphs.append(paragraph)

print("Before deduplication:", len(merged_paragraphs))
print("After deduplication:", len(unique_paragraphs))
```

## Task 6C

```python
stopwords_15 = {
    "the", "is", "a", "an", "and", "or", "to", "of",
    "in", "on", "for", "with", "as", "by", "at"
}

all_words = []

for paragraph in unique_paragraphs:
    clean_paragraph = re.sub(r"[^A-Za-z\s]", " ", paragraph.lower())
    words = clean_paragraph.split()
    words = [word for word in words if word not in stopwords_15]
    all_words.extend(words)

combined_frequency = Counter(all_words)

for word, count in combined_frequency.most_common(20):
    print(word, count)
```

## Task 6D

```python
paragraph_lengths = [len(paragraph.split()) for paragraph in unique_paragraphs]

total_headings = len(headings_a) + len(headings_b)
total_unique_body = len(unique_paragraphs)
vocabulary_size = len(combined_frequency)
average_paragraph_length = sum(paragraph_lengths) / len(paragraph_lengths) if paragraph_lengths else 0
longest_paragraph = max(unique_paragraphs, key=lambda x: len(x.split())) if unique_paragraphs else ""

print("Total headings across both documents:", total_headings)
print("Total unique body paragraphs:", total_unique_body)
print("Vocabulary size:", vocabulary_size)
print("Average paragraph length:", round(average_paragraph_length, 2))
print("Longest paragraph:")
print(longest_paragraph)
```

# Task 8

## Task 8A

```python
todos_url = "https://jsonplaceholder.typicode.com/todos"
todos_response = requests.get(todos_url)

if todos_response.status_code != 200:
    print("Error: status code is", todos_response.status_code)
    raise SystemExit

todos = todos_response.json()

if not isinstance(todos, list):
    print("Error: response is not a list.")
    raise SystemExit

if len(todos) == 0:
    print("Error: no todo records found.")
    raise SystemExit

print("Total todo records retrieved:", len(todos))
```

## Task 8B

```python
todo_summary = {}

for todo in todos:
    user_id = todo["userId"]

    if user_id not in todo_summary:
        todo_summary[user_id] = {
            "total": 0,
            "completed": 0,
            "incomplete": 0
        }

    todo_summary[user_id]["total"] += 1

    if todo["completed"]:
        todo_summary[user_id]["completed"] += 1
    else:
        todo_summary[user_id]["incomplete"] += 1

for user_id, data in todo_summary.items():
    check = data["completed"] + data["incomplete"] == data["total"]
    print(user_id, data, check)
```

## Task 8C

```python
completion_rates = {}

for user_id, data in todo_summary.items():
    rate = data["completed"] / data["total"] * 100
    completion_rates[user_id] = round(rate, 1)

highest_user = max(completion_rates, key=completion_rates.get)
lowest_user = min(completion_rates, key=completion_rates.get)
overall_average = sum(completion_rates.values()) / len(completion_rates)

print("Highest completion rate:", highest_user, completion_rates[highest_user])
print("Lowest completion rate:", lowest_user, completion_rates[lowest_user])
print("Overall average completion rate:", round(overall_average, 1))
```

## Task 8D

```python
users_response_8 = requests.get("https://jsonplaceholder.typicode.com/users")
users_8 = users_response_8.json()

user_lookup = {
    user["id"]: user["name"]
    for user in users_8
}

ranking = sorted(
    completion_rates.items(),
    key=lambda x: x[1],
    reverse=True
)

print(f"{'Rank':<6}{'Name':<25}{'Total Tasks':<14}{'Completed':<12}{'Completion Rate %':<18}")

for rank, (user_id, rate) in enumerate(ranking, start=1):
    data = todo_summary[user_id]
    name = user_lookup[user_id]

    print(
        f"{rank:<6}{name:<25}{data['total']:<14}{data['completed']:<12}{rate:<18}"
    )
```

# Task 9

## Task 9A

```python
users_response_9 = requests.get("https://jsonplaceholder.typicode.com/users")
users_9 = users_response_9.json()

def flatten_user(user: dict) -> dict:
    return {
        "id": user["id"],
        "name": user["name"],
        "username": user["username"],
        "email": user["email"],
        "city": user["address"]["city"],
        "zipcode": user["address"]["zipcode"],
        "lat": user["address"]["geo"]["lat"],
        "lng": user["address"]["geo"]["lng"],
        "company_name": user["company"]["name"],
        "company_catchphrase": user["company"]["catchPhrase"]
    }

flattened_users = [flatten_user(user) for user in users_9]

print(flattened_users)
```

## Task 9B

```python
users_df = pd.DataFrame(flattened_users)

users_df["lat"] = users_df["lat"].astype(float)
users_df["lng"] = users_df["lng"].astype(float)

users_df["distance_from_origin"] = users_df.apply(
    lambda row: math.sqrt(row["lat"] ** 2 + row["lng"] ** 2),
    axis=1
)

print(users_df.to_string(index=False))

greatest_distance_user = users_df.loc[
    users_df["distance_from_origin"].idxmax()
]

print("User with greatest distance from origin:")
print(greatest_distance_user)
```

## Task 9C

```python
northern_users = users_df[users_df["lat"] > 0]
southern_users = users_df[users_df["lat"] < 0]

print("Northern Hemisphere:")
print(northern_users["name"].tolist())

print("Southern Hemisphere:")
print(southern_users["name"].tolist())

north_average = northern_users["distance_from_origin"].mean()
south_average = southern_users["distance_from_origin"].mean()

if north_average > south_average:
    print("Northern Hemisphere has the higher average distance.")
else:
    print("Southern Hemisphere has the higher average distance.")
```
