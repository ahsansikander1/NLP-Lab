# Task 1

`Regex Fundamentals Character Classes and Quantifiers`

```python
import re

text = "Ahsan bought 3 Books in 2026 for Rs. 4500. NLP is Fun! Room 12 has 25 Students."

digits = re.findall(r"\d", text)
capital_words = re.findall(r"\b[A-Z][A-Za-z]*\b", text)
four_digits = re.findall(r"\b\d{4}\b", text)
words_3_to_6 = re.findall(r"\b[A-Za-z]{3,6}\b", text)

print("All digits:", digits)
print("Words starting with a capital letter:", capital_words)
print("Exactly 4 digit sequences:", four_digits)
print("Words with 3 to 6 characters:", words_3_to_6)
```

# Task 2

`Pattern Based Extraction Real World Data`

```python
text = "Customer Ahsan can be reached at ahsan@gmail.com or 0301-1234567. Order ID ORD-2026-101 was placed on 25/09/2026 with a 45% discount. Sara used sara.khan@example.com and phone 0333-7654321 for order ORD-2026-102 on 26/09/2026 and received 12.5% off."

emails = re.findall(r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}", text)
phones = re.findall(r"\b03\d{2}-\d{7}\b", text)
dates = re.findall(r"\b\d{2}/\d{2}/\d{4}\b", text)
percentages = re.findall(r"\d+(?:\.\d+)?%", text)

print("Email addresses:", emails)
print("Phone numbers:", phones)
print("Dates:", dates)
print("Percentages:", percentages)
```

# Task 3

`Word Tokenization NLTK vs spaCy`

```python
import nltk
import spacy
from nltk.tokenize import word_tokenize


paragraph = "Dr. Ali doesn't like delays. It's 10:30 a.m., and he can't wait. Email him at ali@example.com. Visit https://example.com today!"

nltk_tokens = word_tokenize(paragraph)

nlp_blank = spacy.blank("en")
doc = nlp_blank(paragraph)
spacy_tokens = [token.text for token in doc]

print("NLTK tokens:")
print(nltk_tokens)
print("Total NLTK tokens:", len(nltk_tokens))

print("spaCy tokens:")
print(spacy_tokens)
print("Total spaCy tokens:", len(spacy_tokens))
```

#  Comparison

`Difference 1: spaCy keeps ali@example.com as one token while NLTK splits the email into separate parts.`

`Difference 2: spaCy keeps https://example.com as one token while NLTK splits the URL into multiple parts.`

# Task 4

`Token Attributes Categorizing Tokens`

```python
sentence = "Sara paid $50 for 3 books and emailed sara@example.com."

doc = nlp_blank(sentence)

for token in doc:
    print(token.text, "is_alpha:", token.is_alpha, "like_num:", token.like_num, "like_email:", token.like_email)

currency_tokens = [token.text for token in doc if token.is_currency]

print("Currency tokens:", currency_tokens)
```

# Task 5

`Sentence Segmentation Handling Tricky Cases`

```python
nlp = spacy.load("en_core_web_sm")

paragraph = "Dr. Ali works at a hospital in Islamabad. He treats patients every day. Mr. Khan visits him often. They discuss medicine, e.g. new treatments, during meetings."

doc = nlp(paragraph)
sentences = list(doc.sents)

for sentence in sentences:
    print(sentence.text)

print("Total sentences:", len(sentences))
print("Observation: The abbreviations did not cause an incorrect sentence break.")
```
